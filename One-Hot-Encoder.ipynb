```python
# we also do the same here train_test_split method:-

X_train, X_test, y_train, y_test = train_test_split( 
                                     df.drop('selling_price', axis=1), 
                                     df['selling_price'], 
                                     test_size=0.2, 
                                     random_state=0  ) 
```

```python
# now import OneHotEncoder class 

from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(drop='first',sparse_output=False,dtype=np.int32)

'''
  here, drop='first' -> it removes the first two columns from 'fuel' and 'owner' 
  and, sparse_output=False -> by using there is no need to use .toarray() for converting into numpy arrays
  and, dtype=np.int32 -> it is used to convert float values into integer values
'''

X_train_new = ohe.fit_transform(X_train[['fuel','owner']])
X_test_new = ohe.transform(X_test[['fuel','owner']])
```

# OHE with Top categories

```python
counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold=100
```

```python
# make a variable "repl" and store the value which is less than threshold

repl = counts[counts <= threshold].index   # "index" prints the names of brand which has value > threshold
```

```python
# it gives an output in true/false for columns

pd.get_dummies(df['brand'].replace(repl, 'uncommom'))
```

```python
# for getting o/p respone in 0/1, we have to add "dtype=int" above

pd.get_dummies(df['brand'].replace(repl, 'uncommon'), dtype=int)
```
